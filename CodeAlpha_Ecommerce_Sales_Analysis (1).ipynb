```python
import pandas as pd
df = pd.read_csv("/content/sample_data/mnist_train_small.csv")
print("Dataset reloaded successfully!")
display(df.head())
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("duplicates removed.")
print("New shape:", df.shape)
```

```python
print(df.describe())
```

```python
for column in df.columns:
  print("\n", column)
  print(df[column].unique())
```

```python
for column in df.select_dtypes(include='object').columns:
    print("\nValue counts for:",column)
    print(df[column].value_counts())
```

```python
for column in df.select_dtypes(include='object').columns:
    print("\nValue counts for:",column)
    print(df[column].value_counts())
```

```python
import numpy as np
numeric_columns = df.select_dtypes(include=np.number).columns
for column in numeric_columns:
    df[column]=df[column].fillna(df[column].median())
```

```python
catrgorical_columns = df.select_dtypes(include='object').columns
for column in catrgorical_columns:
    df[column]=df[column].fillna(df[column].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
correlation = df.select_dtypes(include=np.number).corr()
print(correlation)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10, 6))
sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap')
plt.show()
```

```python
df.select_dtypes(include=np.number).hist(figsize=(10, 6),bins=20)
plt.suptitle("Distribution of Number Variables")
plt.show()
```

```python
variance = df.var()
columns_to_drop = variance[variance == 0].index
df_filtered = df.drop(columns=columns_to_drop)

print(f"Original number of columns: {df.shape[1]}")
print(f"Number of columns with zero variance dropped: {len(columns_to_drop)}")
print(f"New number of columns: {df_filtered.shape[1]}")

df = df_filtered.copy() # Update df with the filtered DataFrame
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns
for column in numeric_columns:
    plt.figure(figsize=(6, 4))
    sns.boxplot(x=df[column])
    plt.title("Box plot - " + str(column))
    plt.show()
```

```python
plt.figure(figsize=(7,5))
sns.scatterplot(data=df,x=numeric_columns[0],y=numeric_columns[1])
plt.title("Relationship Between Variables")
plt.show()
```

```python
for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR
    outliers = df[(df[column] <= lower_limit) | (df[column] >= upper_limit)]
    print(column, "Outliers:",len(outliers))
```

```python
print("Final Dataset Shape:",df.shape)
print("\nFinal Dataset Information:")
print(df.info())
print("\nFinal Statistical Summary:")
print(df.describe())
```
